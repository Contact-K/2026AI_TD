## (3) 自分の写真ならどうするか

演習2（`ex2.ipynb`）の結果を写す．

| モデル | 平均 ± 標準偏差（%） | CNN との 差 ÷ 標準偏差 と判定 |
|---|---|---|
| CNN（ゼロから） | 84.0 ± 6.4 | ― |
| ResNet18 凍結 | 98.7 ± 1.8 | +14.7 ÷ 4.7 = 3.1 → 差がある |
| ResNet18 微調整 | 100.0 ± 0.0 | +16.0 ÷ 4.5 = 3.5 → 差がある |
| ViT 凍結 | 98.7 ± 1.8 | +14.7 ÷ 4.7 = 3.1 → 差がある |

- 自分の写真は，「写真が多い・少ない」×「ImageNet に似ている・似ていない」のどれに当たるか．理由：少ない × やや似ている。200 枚（訓練 140）で、基板とケーブルという物の写真。顕微鏡画像のように種類が違うわけではない。
- ResNet18 なら，何層まで学習し直すのがよいと考えるか．理由：最後の全結合（fc）だけ。凍結の平均 98.7% に対し、全部を学ぶ微調整は 100% だが差は 1.3 ポイント。学ぶ数は 1,026 個と約 1,118 万個。
- それを確かめるには，どう実験するか（学習する層・回数・何で判定するか）：fc だけ、layer4 と fc、全部、の 3 つをシード 0〜4 で回す。平均の差 ÷ 標準偏差が 2 以上なら「差がある」と判定する。
